```python
from scipy.stats import linregress



def plot_visible_energy_fraction_binned(true_energies, predicted_energies, visible_energies, num_bins=10):
    bins = np.linspace(0, 1, num_bins + 1)
    bin_centers = 0.5 * (bins[:-1] + bins[1:])
    
    fig, ax = plt.subplots(2,5, figsize=(27, 12), sharex=True, sharey=True)
    ax = ax.flatten()
    
    for i in range(num_bins):
        bin_mask = (visible_energies / true_energies > bins[i]) & (visible_energies / true_energies <= bins[i + 1])
        if np.sum(bin_mask) == 0:
            continue
        ax[i].scatter(true_energies[bin_mask], predicted_energies[bin_mask], c=visible_energies[bin_mask]/true_energies[bin_mask], cmap='Spectral', vmin=0, vmax=1., alpha=0.7, s=1, label=f'VE Fraction: {bins[i]:.2f}-{bins[i+1]:.2f}')
    
        y_eq_x = np.linspace(0, 2010, 2010)
        ax[i].plot(y_eq_x, y_eq_x, 'k' '--', alpha=0.7, linewidth=3, label='Perfect Prediction')
        ax[i].set_xlim(-5, 2010)
        ax[i].set_ylim(-500, 4000)
        ax[i].set_xlabel('True KE [MeV]', fontsize=16)
        ax[i].set_ylabel('Predicted Energy [MeV]', fontsize=16)
        ax[i].legend(loc='upper right', fontsize=16)
        ax[i].tick_params(axis='both', which='major', labelsize=14)

        print(len(true_energies[bin_mask]), f'events in VE fraction bin {bins[i]:.2f}-{bins[i+1]:.2f}')
        print(len(predicted_energies[bin_mask]), f'events in VE fraction bin {bins[i]:.2f}-{bins[i+1]:.2f}') 
        x = true_energies[bin_mask]
        y = predicted_energies[bin_mask]
        y = y.flatten()
        #print(x.shape, y.shape)
        res = linregress(x,y)
        fitted_line = res.slope * y_eq_x + res.intercept
        ax[i].plot(y_eq_x, fitted_line, color='rebeccapurple', label='Linear Best Fit', alpha=0.9, linewidth=3)
        ax[i].text(0.05, 0.7, f'Slope: {res.slope:.2f}\nIntercept: {res.intercept:.2f}\nR²: {res.rvalue**2:.2f}', transform=ax[i].transAxes, fontsize=14, verticalalignment='top', bbox=dict(boxstyle='round', facecolor='white', alpha=0.5))

        ax[i].legend(loc='upper right', fontsize=16)
    #cbar_ax = fig.add_axes([0.05, 0.05, 0.9, 0.02]) 
    #cbar = fig.colorbar(ax[0].collections[0], cax=cbar_ax, orientation='horizontal')
    #cbar.set_label('Visible Energy Fraction', fontsize=16)
    fig.suptitle('Sheep CNN Energy Prediction by Visible Energy Fraction Bins', size=20)
    fig.tight_layout()
    plt.show()

plot_visible_energy_fraction_binned(tester.labels, tester.predictions, tester.visible_energy, num_bins=10)
```

```python
from scipy.optimize import curve_fit
def gaussian(x, amplitude, mean, std_dev):
    return amplitude * np.exp(-((x - mean) ** 2) / (2 * std_dev ** 2))


plt.rcParams.update({'font.size': 12})
plt.rcParams.update({'legend.fontsize': 12})
plt.rcParams['font.family'] = 'serif'
#plt.rcParams['mathtext.fontset'] = 'stix'
plt.rcParams['xtick.direction'] = 'in'
plt.rcParams['ytick.direction'] = 'in'
plt.rcParams['xtick.top'] = True
plt.rcParams['ytick.right'] = True
plt.rcParams['xtick.major.size'] = 5
plt.rcParams['xtick.minor.size'] = 3
plt.rcParams['ytick.major.size'] = 5
plt.rcParams['ytick.minor.size'] = 3

res_vis_true = (tester.visible_energy-tester.labels) / tester.labels
res_true_pred = (tester.predictions.flatten()-tester.labels)/ tester.labels
rbins = np.linspace(-10, 10, 201)
num_events=len(res_vis_true)


hist_counts_vis_true, bin_edges_vis_true = np.histogram(res_vis_true, bins=rbins, density=False)
bin_centers_vis_true = (bin_edges_vis_true[:-1] + bin_edges_vis_true[1:]) / 2
initial_guesses_vis_true  = [np.max(hist_counts_vis_true), np.mean(res_vis_true), np.std(res_vis_true)]

hist_counts_true_pred, bin_edges_true_pred = np.histogram(res_true_pred, bins=rbins, density=False)
bin_centers_true_pred = (bin_edges_true_pred[:-1] + bin_edges_true_pred[1:]) / 2
initial_guesses_true_pred  = [np.max(hist_counts_true_pred/num_events), np.mean(res_true_pred), np.std(res_true_pred)]

# Perform the curve fit
vis_true_params, vis_true_covariance = curve_fit(gaussian, bin_centers_vis_true, hist_counts_vis_true, p0=initial_guesses_vis_true)
true_pred_params, true_pred_covariance = curve_fit(gaussian, bin_centers_true_pred, hist_counts_true_pred/num_events, p0=initial_guesses_true_pred)

plt.hist(bin_edges_vis_true[:-1], bins=bin_edges_vis_true, weights=hist_counts_vis_true/num_events, label="(Visible - True) / True", alpha=0.5, edgecolor="none")
#plt.plot(bin_centers_vis_true, gaussian(bin_centers_vis_true, *vis_true_params), color='blue', linestyle="--")
plt.hist(bin_edges_true_pred[:-1],bins=bin_edges_true_pred, weights=hist_counts_true_pred/num_events, label="(SHEEP - True) / True", alpha=0.5, edgecolor="none")
plt.plot(bin_centers_true_pred, gaussian(bin_centers_true_pred, *true_pred_params), color='sienna', linestyle="--")
plt.legend(fontsize=11)
plt.xlim(-2, 2.3)
#plt.ylim(0,0.3)
plt.ylabel("Fraction of Test Events / 0.1")
plt.xlabel("Test Event Energy Resolution")
ax=plt.gca()
ax.text(0.2, 0.2, r"$\mathbf{\mu:}$"+f"{true_pred_params[1]:.2f}\n"+r"$\mathbf{\sigma:}$"+f"{true_pred_params[2]:.2f}", fontsize=12, verticalalignment='top', color='sienna')
ax.text(-1.94, 0.267, f"PRELIMINARY", fontsize=18, verticalalignment='top', color='black', alpha=0.35, fontweight='bold')

plt.savefig("/global/cfs/cdirs/dune/users/ehinkle/nd_prototypes_ana/sheep_fig_chamberlain_fancy_nolines.png")
```

```python
def plot_ve_vs_pred_resolution_binned(true_energies, predicted_energies, visible_energies, num_bins=10):
    ebins = np.linspace(0, 2000, num_bins + 1)
    ebin_centers = 0.5 * (ebins[:-1] + ebins[1:])
    
    fig, ax = plt.subplots(4,5, figsize=(29, 24))#, sharex=True, sharey=True)
    ax = ax.flatten()
    rbins = np.linspace(-10, 10, 201)
    
    for i in range(num_bins):
        bin_mask = (true_energies > ebins[i]) & (true_energies <= ebins[i + 1])
        if np.sum(bin_mask) == 0:
            continue
        ve_bin = visible_energies[bin_mask]
        true_bin = true_energies[bin_mask]
        pred_bin =  predicted_energies[bin_mask]
        pred_bin = pred_bin.flatten()
        res_vis_true = (ve_bin-true_bin) / true_bin
        res_true_pred = (pred_bin-true_bin)/ true_bin

        hist_counts_vis_true, bin_edges_vis_true = np.histogram(res_vis_true, bins=rbins, density=False)
        bin_centers_vis_true = (bin_edges_vis_true[:-1] + bin_edges_vis_true[1:]) / 2
        initial_guesses_vis_true  = [np.max(hist_counts_vis_true), np.mean(res_vis_true), np.std(res_vis_true)]

        hist_counts_true_pred, bin_edges_true_pred = np.histogram(res_true_pred, bins=rbins, density=False)
        bin_centers_true_pred = (bin_edges_true_pred[:-1] + bin_edges_true_pred[1:]) / 2
        initial_guesses_true_pred  = [np.max(hist_counts_true_pred), np.mean(res_true_pred), np.std(res_true_pred)]
        true_pred_params, true_pred_covariance = curve_fit(gaussian, bin_centers_true_pred, hist_counts_true_pred, p0=initial_guesses_true_pred)
        
        ax[i].hist(bin_edges_vis_true[:-1], bins=rbins, weights=hist_counts_vis_true, label=f'(Visible-True)/True', alpha=0.5)
        ax[i].hist(bin_edges_true_pred[:-1], bins=rbins, weights=hist_counts_true_pred, label=f'(Predicted-True)/True', alpha=0.5)
        ax[i].plot(bin_centers_true_pred, gaussian(bin_centers_true_pred, *true_pred_params), color='sienna', linestyle="--")
        #y_eq_x = np.linspace(0, 2010, 2010)
        #ax[i].plot(y_eq_x, y_eq_x, 'k' '--', alpha=0.7, linewidth=3, label='Perfect Prediction')
        ax[i].set_xlim(-2, 3)
        #ax[i].set_ylim(-5, 3000)
        ax[i].set_xlabel('Event Energy Resolution', fontsize=16)
        ax[i].set_ylabel('Count / 0.1', fontsize=16)
        ax[i].legend(loc='upper right', fontsize=16)
        ax[i].tick_params(axis='both', which='major', labelsize=14)
        ax[i].text(0.63, 0.8, f"{ebins[i]:.0f}-{ebins[i+1]:.0f} MeV \nMean: {true_pred_params[1]:.2f} \nStd Dev: {true_pred_params[2]:.2f}", transform=ax[i].transAxes, fontsize=14, verticalalignment='top', bbox=dict(boxstyle='round', facecolor='white', color='orange', alpha=0.2))

        ax[i].legend(loc='upper right', fontsize=16)
    #cbar_ax = fig.add_axes([0.05, 0.05, 0.9, 0.02]) 
    #cbar = fig.colorbar(ax[0].collections[0], cax=cbar_ax, orientation='horizontal')
    #cbar.set_label('Visible Energy Fraction', fontsize=16)
    #fig.suptitle('Sheep CNN Predicted and Visible Energy Resolutions by True Energy Bin', size=20)
    fig.tight_layout()
    plt.savefig("/global/cfs/cdirs/dune/users/ehinkle/nd_prototypes_ana/sheep-model/models/cnn/train_sheep_cnn_nersc/energy_res.png")

plot_ve_vs_pred_resolution_binned(tester.labels, tester.predictions, tester.visible_energy, num_bins=20)
```
